```python
import os
import logging
import matplotlib.pyplot as plt
import networkx as nx
from typing import Dict, List, Any
import sys
from dotenv import load_dotenv


sys.path.append(os.path.abspath(".."))


from classification.concept_minor_classifier import ConceptMinorClassifier
from common.params import ClassificationParams

load_dotenv()


# Configure logging
logging.basicConfig(level=logging.INFO, 
                   format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')
```

```python
params = ClassificationParams(
    feature_weight=0.5,
    structural_weight=0.5,
    ged_timeout=0.5,
)

logging.info(f"Classification params: {params}")
logging.info(f"Neo4j DSN: {os.getenv('NEO4J_DSN')}")
logging.info(f"Neo4j User: {os.getenv('NEO4J_USER')}")
logging.info(f"Neo4j Pass: {os.getenv('NEO4J_PASSWORD')}")


graph_comparator = ConceptMinorClassifier(
    neo4j_dsn=os.getenv("NEO4J_DSN"),
    neo4j_user=os.getenv("NEO4J_USER"),
    neo4j_pass=os.getenv("NEO4J_PASSWORD"),
    max_workers=10,
    use_multithreading=False,
)
```

```python
image_id = "807a98c6-e700-49d8-9826-c918b7f0ce15"
results = graph_comparator.classify(image_id, params)
graph_comparator.close()
results
```
