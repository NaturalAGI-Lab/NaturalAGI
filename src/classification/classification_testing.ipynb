# Classification testing notebook

```python
import os
import logging
import sys
from dotenv import load_dotenv


sys.path.append(os.path.abspath(".."))


from classification.concept_minor_classifier import ConceptMinorClassifier
from classification.models import ClassificationResult
from classification.repository.concept_repository import ConceptRepository
from classification.repository.image_repository import ImageRepository

load_dotenv()


# Configure logging
logging.basicConfig(
    level=logging.INFO, format="%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)
```

```python
image_repository = ImageRepository(
    neo4j_dsn=os.getenv("NEO4J_DSN"),
    neo4j_user=os.getenv("NEO4J_USER"),
    neo4j_pass=os.getenv("NEO4J_PASSWORD"),
)

concept_repository = ConceptRepository(
    neo4j_dsn=os.getenv("NEO4J_DSN"),
    neo4j_user=os.getenv("NEO4J_USER"),
    neo4j_pass=os.getenv("NEO4J_PASSWORD"),
)

graph_comparator = ConceptMinorClassifier(
    concept_repository,
    image_repository,
    max_workers=2,
    use_multithreading=False,
    ged_timeout=15,
)
```

```python
image_id = "00b1b8e7-bcd7-428e-9bee-aa37c0db9101"
results = graph_comparator.classify(image_id)
image_repository.close()
concept_repository.close()
print("\n".join([str(result) for result in results]))
```
