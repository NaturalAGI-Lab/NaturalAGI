```python
# Detect lines using the line_detector.py

import os
import cv2
from line_detector import detect_lines
import matplotlib.pyplot as plt
import numpy as np

# Set the folder path containing the images
folder_path = "../../tests/generated_samples/mnist_ones"

# Get a list of image files in the folder
image_files = [f for f in os.listdir(folder_path) if f.endswith(('.bmp', '.jpg', '.png'))]

# Function to display an image with detected lines
def display_image_with_lines(image: np.ndarray, lines: list) -> None:
    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    for line in lines:
        plt.plot([line['x1'], line['x2']], [line['y1'], line['y2']], 'r-')
    plt.axis('off')
    plt.show()

# Process each image in the folder
for image_file in image_files:
    image_path = os.path.join(folder_path, image_file)
    image = cv2.imread(image_path)
    
    if image is None:
        print(f"Failed to load image: {image_file}")
        continue
    
    lines = detect_lines(image)
    print(f"Image: {image_file}")
    print(f"Number of lines detected: {len(lines)}")
    print("Lines:", lines)
    print("\n")
    
    # Display the image with detected lines
    display_image_with_lines(image, lines)
```
