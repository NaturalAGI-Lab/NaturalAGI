```python
# Detect lines using the line_detector.py

import os
import cv2
from line_detector import detect_lines, skeletonize_image
import matplotlib.pyplot as plt
import numpy as np

# Set the folder path containing the images
folder_path = "../../tests/generated_samples/mnist_fours"

# Get a list of image files in the folder
image_files = [f for f in os.listdir(folder_path) if f.endswith(('.bmp', '.jpg', '.png'))]

# Function to display an image with detected lines
def display_images_with_lines(original: np.ndarray, skeleton: np.ndarray, lines: list) -> None:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 10))
    
    # Original image
    ax1.imshow(cv2.cvtColor(original, cv2.COLOR_BGR2RGB))
    ax1.set_title("Original Image")
    ax1.axis('off')
    
    # Skeletonized image
    ax2.imshow(skeleton, cmap='gray')
    ax2.set_title("Skeletonized Image")
    ax2.axis('off')
    
    # Plot lines on both images
    for line in lines:
        ax1.plot([line['x1'], line['x2']], [line['y1'], line['y2']], 'r-')
        ax2.plot([line['x1'], line['x2']], [line['y1'], line['y2']], 'r-')
    
    plt.tight_layout()
    plt.show()

# Process each image in the folder
for image_file in image_files:
    image_path = os.path.join(folder_path, image_file)
    image = cv2.imread(image_path)
    
    if image is None:
        print(f"Failed to load image: {image_file}")
        continue
    
    # Get skeletonized image
    skeleton = skeletonize_image(image)
    
    lines = detect_lines(image)
    print(f"Image: {image_file}")
    print(f"Number of lines detected: {len(lines)}")
    print("Lines:", lines)
    print("\n")
    
    # Display the original and skeletonized images with detected lines
    display_images_with_lines(image, skeleton, lines)
```
