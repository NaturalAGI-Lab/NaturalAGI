```python
import json
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from typing import Dict, Any

def visualize_networkx_json_graph(json_data: str) -> None:
    """
    Visualize a NetworkX JSON graph using matplotlib.

    Args:
        json_data (str): JSON string representing the NetworkX graph.

    Returns:
        None
    """
    # Parse JSON data
    graph_data: Dict[str, Any] = json.loads(json_data)

    # Create a new graph
    G = nx.Graph()

    # Add nodes
    for node in graph_data['nodes']:
        node_id = eval(node['id'])  # Convert string tuple to actual tuple
        G.add_node(node_id)

    # Add edges
    for link in graph_data['links']:
        source = eval(link['source'])
        target = eval(link['target'])
        G.add_edge(source, target)

    # Set up the plot
    plt.figure(figsize=(12, 8))
    
    # Extract x and y coordinates from node ids
    pos = {node: (node[0], node[1]) for node in G.nodes()}

    # Draw the graph
    nx.draw(G, pos, with_labels=False, node_size=50, node_color='lightblue', edge_color='gray')

    # Add labels to nodes (optional)
    labels = {node: f"({node[0]:.2f}, {node[1]:.2f})" for node in G.nodes()}
    nx.draw_networkx_labels(G, pos, labels, font_size=8)

    plt.title("NetworkX Graph Visualization")
    plt.axis('off')
    plt.tight_layout()
    plt.show()

visualize_networkx_json_graph(json_data)
```
